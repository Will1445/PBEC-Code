```python
import numpy as np
import matplotlib.pyplot as plt

directory = 'D:/Data/2024/202410/20241030/'

import os
import numpy as np
import re

import os
import numpy as np
import re
 
def load_data(directory):
    Times = {}
    Voltages = {}
    pattern = r"_(\d+(\.\d+)?)_(Time|Voltage)\.txt$"

    for filename in os.listdir(directory):
        if filename.endswith('txt'):
            match = re.search(pattern, filename)
            if match:
                power = float(match.group(1))
                data_type = match.group(3).lower()
                file_path = os.path.join(directory,filename)
                data = np.loadtxt(file_path)

                if data_type == 'time':
                    Times[power] = data
                elif data_type == 'voltage':
                    Voltages[power] = data

    return Times,Voltages


# Example usage
directory = 'D:/Data/2024/202411/20241118'  # Specify the directory containing the files
Times = np.loadtxt('D:/Data/2024/202411/20241118/pbec_100922_100_1_Time.txt')
Voltage1 = np.loadtxt('D:/Data/2024/202411/20241118/pbec_100922_100_1_channel0.txt')
Voltage2 = np.loadtxt('D:/Data/2024/202411/20241118/pbec_100922_100_1_channel1.txt')

# mean_voltages = {}
# for power, voltage_data in Voltages.items():
#     mean_voltages[power] = np.mean(voltage_data)
#     
# powers = sorted(mean_voltages.keys())
# mean_voltages_values = [mean_voltages[power] for power in powers]

plt.plot(Times, Voltage2)
plt.xlabel('power in mW')
plt.ylabel('mean voltage in mV')
```
