```python
from CameraUSB3 import CameraUSB3
import matplotlib.pyplot as plt
import cv2
import numpy as np

cam = CameraUSB3(verbose=True, camera_id='blackfly_semiconductor_cavity_lock', timeout=1000, acquisition_mode='single frame')

image = cam.get_image()

import cv2
import numpy as np
import matplotlib.pyplot as plt


# Define region and shift
x1, x2 = 700, 800
y1, y2 = 600, 700
shift = 150

# Create copy to modify
modified_image = image

# Replace each pixel in the region using value from x-shift
modified_image[y1:y2, x1:x2] = image[y1:y2, x1-shift:x2-shift]

# Plot
plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
plt.title('Original')
plt.imshow(image, cmap='viridis')

plt.subplot(1, 2, 2)
plt.title('Modified')
plt.imshow(modified_image, cmap='viridis')

plt.tight_layout()
plt.show()

```
