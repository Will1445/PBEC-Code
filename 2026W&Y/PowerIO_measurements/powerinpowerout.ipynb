```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import sys
import socket
import time
from datetime import datetime
from pathlib import Path

if socket.gethostname() == "ph-photonbec5":
    sys.path.append(r"D:/Control/PythonPackages/")

import Components as comp

laser = comp.Toptica_Laser()

# Power Meter A only
powermeter = comp.Thor_PowerMeter(
    power_meter_usb_name='USB0::0x1313::0x8078::P0034379::INSTR',
    num_power_readings=15,
    bs_factor=1,
    wavelength=950
)

powermeter2 = comp.Thor_PowerMeter(
    power_meter_usb_name='USB0::0x1313::0x8078::P0046773::INSTR',
    num_power_readings=15,
    bs_factor=1,
    wavelength=950
)
```

```python
laser_powers = np.arange(0,0.04,0.001)
output_powers = []
output_powers2 = []
```

```python

for power in laser_powers:
    laser.set(power=power)
    output_power = powermeter.take_power_reading()
    output_power2 = powermeter2.take_power_reading()
    output_powers.append(output_power)
    output_powers2.append(output_power2)
    print(f"measured power at: {power}")
```

```python
import matplotlib.pyplot as plt
# plt.plot(new_laser_powers, new_output_powers, 'x')
# plt.show()

total_powers = [a + b for a, b in zip(output_powers, output_powers2)]
import numpy as np

x = np.asarray(laser_powers)

for powers, label in [
    (output_powers, "PM1"),
    (output_powers2, "PM2"),
    (total_powers, "PM1+PM2"),
]:
    y = np.asarray(powers)
    mask = y < 0.0001
    plt.plot(x[mask], y[mask], "x", label=label)

plt.legend()
plt.legend()
plt.show()
```
